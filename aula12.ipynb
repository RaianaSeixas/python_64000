### Automatizar a execução de tarefas com schedule

```python
#!pip install schedule
import schedule
import time
```

```python
def tarefa():
    print('Test schedule\n')

job_tarefa = schedule.every(2).seconds.do(tarefa)

start_time = time.time()  # Armazena o tempo inicial

while True:
    schedule.run_pending()  # Verifica todas as tarefas programadas
    time.sleep(1)  # Pausa por 1 segundo
    
    elapsed_time = time.time() - start_time  # Calcula o tempo decorrido
    
    if elapsed_time > 10:  # Verifica se já passaram 10 segundos
        break  # Sai do loop

print("Loop finalizado após 10 segundos.")
```

```python
# Verificando todas as tarefas pendentes
for job in schedule.jobs:
    print(f"Tarefa: {job.job_func.__name__}, Intervalo: {job.interval} {job.unit}, Próxima execução: {job.next_run}")
```

```python
def obter_info_aleatoria_wikipedia():
    url = "https://pt.wikipedia.org/api/rest_v1/page/random/summary"
    resposta = requests.get(url)

    if resposta.status_code == 200:
        dados = resposta.json()
        print(f"Título: {dados['title']}")
        print(f"Resumo: {dados['extract']}\n")
    else:
        print("Erro ao acessar a Wikipédia")

# Agendar a função para ser chamada a cada 5 segundos
job_wikipedia = schedule.every(1).seconds.do(obter_info_aleatoria_wikipedia)

start_time = time.time()
while True:
    schedule.run_pending()  # Verifica todas as tarefas programadas
    time.sleep(1)  # Pausa por 1 segundo
    
    elapsed_time = time.time() - start_time  # Calcula o tempo decorrido
    
    if elapsed_time > 15:  # Verifica se já passaram 10 segundos
        break  # Sai do loop

print("Loop finalizado.")
```

```python
# Verificando todas as tarefas programadas

# schedule.run_pending() roda todas as tarefas ativas que foram programadas.

for job in schedule.jobs:
    print(f"Tarefa: {job.job_func.__name__}, Intervalo: {job.interval} {job.unit}, Próxima execução: {job.next_run}")
```

```python
# Cancelando a função 'tarefa' que estava programada

schedule.cancel_job(job_tarefa)  # Cancela o job específico
```
