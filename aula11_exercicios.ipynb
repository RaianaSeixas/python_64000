```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_excel(r'dados/fastfood.xlsx')
df.head(2)
```

```python
df1 = df.groupby('restaurant')[['restaurant']].count().rename(columns={'restaurant':'count'}).reset_index()
df1
```

```python
sns.barplot(x='restaurant', y='count', data = df1)
```

```python
sns.barplot(x='restaurant', y='count', data = df1, palette='Blues', edgecolor="black")
plt.title('Contagem de itens por restaurante')
plt.xticks(rotation=30)
plt.grid(False)
# Iterate through each container and add labels
for container in plt.gca().containers:
    plt.bar_label(container)
plt.show()
```

```python
df.hist(figsize=(10,8))
plt.tight_layout()
plt.show()
```

```python
sns.histplot(df['calories'], kde = True)
plt.show()
```

```python
sns.histplot(df['total_fat'], kde = True)
plt.show()
```

```python
sns.set_theme(style="white",rc={'figure.figsize':(10, 6)})
sns.histplot(df['protein'], kde=True, color='red', label='protein')
sns.histplot(df['total_fat'], kde=True, multiple="layer", label='total_fat')
plt.legend()
plt.xlabel('Quantidade')
plt.ylabel('Contagem')
plt.title('Histograma proteína vs gordura total')
plt.show()
```

```python
sns.set_theme(rc={'figure.figsize':(8,5)})
sns.scatterplot(x='protein', y='total_fat', data = df)
plt.show()
```

```python
sns.scatterplot(x='calories', y='total_fat', data = df)
```

```python
sns.set_theme(rc={'figure.figsize':(10,8)})
sns.boxplot(hue = 'restaurant', y='total_fat', data = df, palette='pastel') #'Set2','coolwarm',"pastel", "husl", "coolwarm" 
plt.grid(True)
plt.show()
```
